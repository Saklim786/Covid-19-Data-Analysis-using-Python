```python
# 16. Filter the dataframe with only this columns: ['continent','location','date','total_cases','total_deaths','gdp_per_capita','human_development_index'] and update the data frame.


df=df[['continent','location','date','total_cases','total_deaths','gdp_per_capita','human_development_index']]
df
```

```python
df
```

```python
# 17. Data Cleaning
# a). Remove all duplicates observations


df.duplicated().sum()
```

```python
# 18.b). Find missing values in all columns


df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
# 19.c). Remove all observations where continent column value is missing


df.dropna(subset=["continent"])
```

```python
# 20.d). Fill all missing values with 0


df= df.fillna(0)
```

df.isnull().sum()

```python
# 21.Date time format :




df['date']= pd.to_datetime(df['date'])
```

```python
# 22.a). Convert date column in datetime format using pandas.to_datetime

df['date'] = pd.to_datetime(df['date'], format="%Y-%m-%d")
```

```python
df.dtypes
```

```python
# 23.b). Create new column month after extracting month data from date column


df['month'] = pd.DatetimeIndex(df['date']).month
```

```python
df
```

```python
# 24.Data Aggregation:
# a). Find max value in all columns using groupby function on 'continent' column


df.groupby('continent').max().reset_index()
```

```python
# 25.b). Store the result in a new dataframe named 'df_groupby'. (Use df_groupby dataframe for all further analysis)


df_groupby = df.groupby('continent').max().reset_index()
df_groupby
```

```python
# 26. Feature Engineering :
# a). Create a new feature 'total_deaths_to_total_cases' by ratio of 'total_deaths' column to 'total_cases'


df_groupby["total_deaths_to_total_cases"]=df_groupby["total_deaths"]/df_groupby["total_cases"]
```

```python
df_groupby["total_deaths_to_total_cases"]*100
```

```python
df_groupby
```

```python
# 27.Data Visualization :



sns.distplot(df["gdp_per_capita"])
plt.show()
```

```python
# 28.a). Perform Univariate analysis on 'gdp_per_capita' column by plotting histogram using seaborn dist plot.


sns.histplot(df["gdp_per_capita"], kde=True)
plt.xlabel("GDP per Capita")
plt.ylabel("Density")
plt.title("Distribution of GDP per Capita")
plt.show()
```

```python
# 29.b). Plot a scatter plot of 'total_cases' & 'gdp_per_capita'


sns.jointplot(data=df_groupby,x="total_cases",y="gdp_per_capita",kind="scatter")
plt.show()
```

```python
# 30.c). Plot Pairplot on df_groupby dataset.


sns.pairplot(data=df_groupby)
plt.show()
```

```python
sns.catplot(data=df_groupby,x="continent",y="total_cases",kind="bar")
plt.show()
```
